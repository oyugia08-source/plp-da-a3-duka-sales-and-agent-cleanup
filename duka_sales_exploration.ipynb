Weekend sales generated 28.5% of revenue alongside 705 Premium orders.

## First-Five-Minutes Inspection & Derived Columns

```python
import numpy as np
import pandas as pd

df = pd.read_csv('duka_sales.csv')
df['order_date'] = pd.to_datetime(df['order_date'])
print('Shape:', df.shape)
print('Dtypes:\n', df.dtypes)
print('Missing:\n', df.isna().sum())
print('Date range:', df['order_date'].min(), 'to', df['order_date'].max())
display(df.head())
display(df.describe(include='all'))

df['revenue'] = df['unit_price'] * df['quantity']
df['month'] = df['order_date'].dt.month_name()
df['weekday'] = df['order_date'].dt.day_name()
df['is_weekend'] = df['order_date'].dt.dayofweek >= 5
df['price_band'] = np.where(df['unit_price'] >= 28000, 'Premium', 'Value')
display(df[['order_id', 'revenue', 'month', 'weekday', 'is_weekend', 'price_band']].head())
```

## Task 3: Branch Revenue Ranking

```python
branch_summary = df.groupby('branch').agg(
    total_revenue=('revenue', 'sum'),
    orders=('order_id', 'count'),
    units_sold=('quantity', 'sum'),
).sort_values('total_revenue', ascending=False)
display(branch_summary)
```

### Task 3 Answer

**Nairobi** achieved the highest revenue at **KES 33,462,378.00** across **369** orders, while **Eldoret** had the lowest revenue at **KES 11,664,112.00**.

## Task 4: Product & Price Band Performance

```python
product_band_summary = df.groupby(['product', 'price_band']).agg(
    total_revenue=('revenue', 'sum'),
    orders=('order_id', 'count'),
    avg_unit_price=('unit_price', 'mean'),
).sort_values('total_revenue', ascending=False)
display(product_band_summary)
```

### Task 4 Answer

The top product-price-band segment by revenue is **Laptop (Premium)** with **KES 30,699,995.00** across **193** orders.

## Task 5: Weekend vs. Weekday Sales

```python
weekend_summary = df.groupby('is_weekend').agg(
    total_revenue=('revenue', 'sum'),
    orders=('order_id', 'count'),
    avg_order_value=('revenue', 'mean'),
)
display(weekend_summary)
```

### Task 5 Answer

Weekend orders generated **KES 26,050,111.00** (avg order value **KES 85,410.20**) compared to weekday revenue of **KES 65,445,970.00** (avg order value **KES 94,166.86**).

## Task 6: Monthly Revenue Trend

```python
months = ['January', 'February', 'March', 'April', 'May', 'June']
monthly_summary = df.groupby('month').agg(
    total_revenue=('revenue', 'sum'),
    orders=('order_id', 'count'),
).reindex(months)
display(monthly_summary)
```

### Task 6 Answer

Monthly revenue was highest in **June** (**KES 17,726,999.00**) and lowest in **May** (**KES 12,822,059.00**).

## Task 7: Filtered Export (`outputs/premium_online_coast_capital.csv`)

```python
import os
os.makedirs('outputs', exist_ok=True)
filtered = df[
    (df['price_band'] == 'Premium')
    & (df['channel'] == 'Online')
    & (df['branch'].isin(['Nairobi', 'Mombasa']))
]
filtered.to_csv('outputs/premium_online_coast_capital.csv', index=False)
print('Saved rows:', len(filtered))
display(filtered.head())
```

### Task 7 Answer

Exported **142** Premium, Online orders from Nairobi and Mombasa totaling **KES 17,270,114.00** to `outputs/premium_online_coast_capital.csv` with `index=False`.
